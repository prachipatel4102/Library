```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
```

```python
delivery_times = np.random.randint(15, 60, size=50)  # minutes

plt.figure(figsize=(6, 5))
sns.histplot(delivery_times, bins=10, kde=False, color="pink")
plt.title("Zomato Delivery Times (50 Orders)")
plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Number of Orders")
plt.show()
```

```python
tips = sns.load_dataset("tips")

plt.figure(figsize=(7, 4))
sns.boxplot(data=tips, x="day", y="total_bill")
plt.title("Total Bill by Day of the Week")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
sns.set_theme(style="darkgrid")

teams = ["MI", "CSK", "RCB", "KKR", "DC", "GT", "RR", "PBKS"]
ipl = pd.DataFrame({
    "team": np.repeat(teams, 20),
    "runs": np.random.randint(110, 230, size=8 * 20)
})
plt.figure(figsize=(10, 5))
sns.violinplot(data=ipl, x="team", y="runs")
plt.title("IPL Run Distribution per Team")
plt.xlabel("Team")
plt.ylabel("Runs")
plt.show()
```

```python
genres = np.random.choice(["Pop", "Rock", "Hip-Hop", "Classical", "Jazz"], size=40)
songs = pd.DataFrame({"genre": genres})

plt.figure(figsize=(7, 4))
sns.countplot(data=songs, x="genre", palette="viridis")
plt.title("Number of Songs per Genre")
plt.xlabel("Genre")
plt.ylabel("Number of Songs")
plt.show()
```

```python
sns.set_theme(style="whitegrid")

steps = [6500, 8200, 7100, 9400, 5800, 10200, 7600]

plt.figure(figsize=(7, 4))
sns.kdeplot(steps, color="green", fill=True)
plt.title("Distribution of Daily Step Counts")
plt.xlabel("Steps")
plt.ylabel("Density")
plt.show()
```
