```python
import sweetviz as sv

mumbai = pd.DataFrame({
    "restaurant": [f"M{i}" for i in range(50)],
    "average_cost_for_two": np.random.randint(300, 2500, 50),
    "user_rating": np.random.uniform(3.0, 4.9, 50).round(1),
})
delhi = pd.DataFrame({
    "restaurant": [f"D{i}" for i in range(50)],
    "average_cost_for_two": np.random.randint(200, 1800, 50),
    "user_rating": np.random.uniform(3.2, 4.8, 50).round(1),
})
# or: mumbai = pd.read_csv("zomato_mumbai.csv"), delhi = pd.read_csv("zomato_delhi.csv")

report = sv.compare([mumbai, "Mumbai"], [delhi, "Delhi"])
report.show_html("zomato_compare.html")
```
