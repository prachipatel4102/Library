```python
import pandas as pd
import numpy as np

data = {
    "player": ["Virat Kohli", "Rohit Sharma", "MS Dhoni", "KL Rahul", "Hardik Pandya", "Jasprit Bumrah", "Rashid Khan", "Shubman Gill"],
    "runs": [7263, 5879, 5243, np.nan, 2525, np.nan, 443, 3000],
    "matches": [237, 243, 250, 132, 123, 120, np.nan, 91]
}
pd.DataFrame(data).to_csv("ipl_player_stats.csv", index=False)

df = pd.read_csv("ipl_player_stats.csv")
print(df)

print("\nisnull:")
print(df.isnull().sum())

print("\nnotnull:")
print(df.notnull().sum())
```

```python
print("Before:", df.shape)

df_dropped = df.dropna(axis=0, how="any")
print("After:", df_dropped.shape)
print(df_dropped)
```

```python
mean_runs = df["runs"].mean()
df["runs"] = df["runs"].fillna(mean_runs)

print("Mean:", mean_runs)
print(df["runs"])
```

```python
ratings = pd.DataFrame({
    "restaurant": ["A", "B", "C", "D", "E", "F", "G"],
    "rating": [np.nan, 4.2, np.nan, np.nan, 3.8, np.nan, 4.5]
})
print("Before:")
print(ratings)

ratings["rating"] = ratings["rating"].ffill()
print("\nAfter forward fill:")
print(ratings)

ratings["rating"] = ratings["rating"].bfill()
print("\nAfter backward fill:")
print(ratings)
```

```python
# task 5 
# The missing values in the Zomato `rating` column are most likely MNAR (Missing Not At Random).
# Customers who had a bad experience often do not bother to leave a rating, so the missingness 
# depends on the rating value itself.
```
