```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)
```

```python
ipl = pd.DataFrame({
    "match_id": range(1, 31),
    "total_runs": np.random.randint(110, 240, size=30)
})
# ipl = pd.read_csv("ipl_matches.csv")

print("Mean:  ", ipl["total_runs"].mean())
print("Median:", ipl["total_runs"].median())
print("Min:   ", ipl["total_runs"].min())
print("Max:   ", ipl["total_runs"].max())
print("Std:   ", ipl["total_runs"].std())

# or all at once:
print(ipl["total_runs"].describe())
```

```python
reviews = pd.DataFrame({
    "rating": np.random.choice([1, 2, 3, 4, 5], size=200, p=[0.08, 0.10, 0.17, 0.30, 0.35]),
    "category": np.random.choice(["Mobiles", "Laptops", "Fashion"], size=200)
})

plt.figure(figsize=(6, 4))
sns.countplot(data=reviews, x="rating", order=[1, 2, 3, 4, 5], palette="viridis")
plt.title("Number of Reviews per Rating")
plt.xlabel("Rating (stars)")
plt.ylabel("Count of Reviews")
plt.show()
```

```python
zomato = pd.DataFrame({
    "average_cost_for_two": np.random.randint(200, 2000, size=60)
})
zomato["user_rating"] = (3 + zomato["average_cost_for_two"] / 1500 + np.random.normal(0, 0.3, 60)).clip(1, 5)

plt.figure(figsize=(7, 5))
sns.scatterplot(data=zomato, x="average_cost_for_two", y="user_rating")
plt.title("Cost for Two vs User Rating")
plt.xlabel("Average Cost for Two (₹)")
plt.ylabel("User Rating")
plt.show()
```

```python
movies = pd.DataFrame({
    "genre": np.random.choice(["Action", "Comedy", "Drama", "Romance", "Thriller"], size=50),
    "box_office_collection": np.random.randint(10, 500, size=50)  # in crores
})

avg_box_office = (
    movies.groupby("genre")["box_office_collection"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)
print(avg_box_office)
```

```python
n = 100
danceability = np.random.uniform(0.2, 0.95, n)
spotify = pd.DataFrame({
    "danceability": danceability,
    "energy": (0.5 * danceability + np.random.uniform(0.1, 0.6, n)).clip(0, 1),
    "valence": np.random.uniform(0.05, 0.95, n),
    "popularity": np.random.randint(20, 100, n)
})

sns.pairplot(spotify)
plt.show()
```
