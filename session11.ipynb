```python
import matplotlib.pyplot as plt
```

```python
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# Line
axs[0, 0].plot([1, 2, 3, 4, 5], [2, 4, 3, 6, 5], marker="o")
axs[0, 0].set_title("Line Chart")

# Bar
axs[0, 1].bar(["A", "B", "C", "D"], [10, 15, 7, 12], color="orange")
axs[0, 1].set_title("Bar Chart")

# Scatter
axs[1, 0].scatter([1, 2, 3, 4, 5, 6], [5, 3, 6, 2, 7, 4], color="green")
axs[1, 0].set_title("Scatter Plot")

# Pie
axs[1, 1].pie([30, 25, 20, 25], labels=["W", "X", "Y", "Z"], autopct="%1.0f%%")
axs[1, 1].set_title("Pie Chart")

plt.tight_layout()
plt.show()
```

```python
platforms = ["Zomato", "Swiggy", "Domino's"]
times = [32, 28, 25]  # average delivery time in minutes

colors = ["red", "orange", "blue"]
linestyles = ["-", "--", ":"]
linewidths = [1, 2, 3]

plt.figure(figsize=(6, 4))
bars = plt.bar(platforms, times, color=colors, edgecolor="black")

for bar, ls, lw in zip(bars, linestyles, linewidths):
    bar.set_linestyle(ls)
    bar.set_linewidth(lw)

plt.title("Average Delivery Time")
plt.xlabel("Platform")
plt.ylabel("Minutes")
plt.show()
```

```python
influencers = ["Inf1", "Inf2", "Inf3", "Inf4", "Inf5"]
followers = [120000, 450000, 80000, 300000, 220000]
posts_per_day = [2, 4, 1, 3, 2]

fig, ax1 = plt.subplots(figsize=(8, 5))

ax1.plot(influencers, followers, color="blue", marker="o")
ax1.set_xlabel("Influencer")
ax1.set_ylabel("Followers", color="blue")
ax1.tick_params(axis="y", labelcolor="blue")

ax2 = ax1.twinx()
ax2.plot(influencers, posts_per_day, color="red", marker="s")
ax2.set_ylabel("Average Daily Posts", color="red")
ax2.tick_params(axis="y", labelcolor="red")

plt.title("Followers vs Daily Posts")
plt.show()
```

```python
movies = ["Movie A", "Movie B", "Movie C", "Movie D", "Movie E"]
tickets_sold = [500000, 800000, 350000, 1200000, 650000]
imdb = [7.2, 6.5, 8.1, 7.8, 6.9]

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(tickets_sold, imdb, color="purple", s=80)

for name, x, y in zip(movies, tickets_sold, imdb):
    ax.annotate(name, (x, y), textcoords="offset points", xytext=(0, 10), ha="center")

ax.set_xlabel("Tickets Sold")
ax.set_ylabel("IMDb Rating")
ax.set_title("Bollywood Movies: Tickets vs IMDb Rating")
plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
sales = [40, 35, 42, 45, 48, 50, 47, 52, 60, 95, 70, 65]  # in crores

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(months, sales, marker="o")

max_idx = sales.index(max(sales))
ax.annotate(
    "Big Billion Days",
    xy=(months[max_idx], sales[max_idx]),
    xytext=(max_idx - 3, sales[max_idx] - 10),
    arrowprops=dict(facecolor="red", arrowstyle="->")
)

ax.set_title("Flipkart Monthly Sales")
ax.set_xlabel("Month")
ax.set_ylabel("Sales (crores)")
plt.show()
```
