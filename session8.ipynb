```python
import pandas as pd
```

```python
orders = pd.DataFrame({
    "restaurant": ["Pizza Hut", "Domino's", "Pizza Hut", "KFC", "Domino's", "KFC"],
    "order_date": ["2026-09-10", "2026-09-11", "2026-09-10", "2026-09-12", "2026-09-11", "2026-09-15"]
})
# orders = pd.read_csv("zomato_orders.csv")  # if you use a real CSV

dups = orders[orders.duplicated(subset=["restaurant", "order_date"])]
print("Duplicate orders:")
print(dups)
```

```python
reviews = pd.DataFrame({
    "review": [
        "Great product", "Worst quality", "Great product", "Value for money",
        "Great product", "Worst quality", "Value for money", "Nice", "Worst quality", "Great product"
    ]
})

top3 = reviews["review"].value_counts().head(3)
print(top3)
```

```python
playlists = pd.DataFrame({
    "playlist_name": ["Chill Vibes", "Gym Mix", "Chill Vibes", "Road Trip", "Gym Mix", "Focus"],
    "creator": ["alex", "sam", "alex", "riya", "sam", "john"]
})

clean = playlists.drop_duplicates().reset_index(drop=True)
print(clean)
```

```python
insta = pd.DataFrame({
    "username": ["insta_queen", "insta-queen", "instaqueen", "insta_queen", "cool_boy", "insta-queen"]
})

insta["username"] = insta["username"].replace(
    ["insta_queen", "insta-queen"], "instaqueen"
)
print(insta)
```

```python
paytm = pd.DataFrame({
    "status": ["Yes", "yes", "Y", "No", "no", "N", " yes ", "N ", " Y"]
})

paytm["status"] = (
    paytm["status"]
    .str.strip()
    .str.lower()
    .map({"yes": 1, "y": 1, "no": 0, "n": 0})
)
print(paytm)
```
