```python
import pandas as pd
import numpy as np
```

```python
timestamps = ['2024-06-01 14:30', '2024-06-02 09:15', '2024-06-03 20:45']

dt = pd.to_datetime(timestamps)
print(dt)
```

```python
orders = pd.DataFrame({
    "order_date": [
        "2024-06-01 14:30:00", "2024-06-02 09:15:00", "2024-06-03 20:45:00",
        "2024-06-08 11:00:00", "2024-06-10 18:20:00", "2024-06-15 13:10:00",
        "2024-06-17 21:05:00", "2024-06-23 10:40:00"
    ]
})
# orders = pd.read_csv("orders.csv")  # if you use a real CSV

orders["order_date"] = pd.to_datetime(orders["order_date"], format="%Y-%m-%d %H:%M:%S")
orders["year"] = orders["order_date"].dt.year
orders["month"] = orders["order_date"].dt.month
orders["weekday"] = orders["order_date"].dt.day_name()
print(orders)
```

```python
orders_idx = orders.set_index("order_date")
weekly_orders = orders_idx.resample("W").size()
print(weekly_orders)
```

```python
posts = pd.DataFrame({
    "posted_at": pd.date_range("2024-06-01 00:00", periods=8, freq="7h")
})

posts["posted_at"] = posts["posted_at"].dt.tz_localize("UTC")
posts["posted_at_ist"] = posts["posted_at"].dt.tz_convert("Asia/Kolkata")
print(posts["posted_at_ist"].head())
```

```python
orders["is_weekend"] = orders["order_date"].dt.dayofweek >= 5
print(orders)
```
