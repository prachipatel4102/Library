```python
import pandas as pd
from sqlalchemy import create_engine, text

engine = create_engine("mysql+pymysql://root:@localhost:3306/zomato_db")

query = "SELECT name, rating FROM movies WHERE rating > 8"
df = pd.read_sql_query(text(query), engine)
print(df)
```

```python
import pandas as pd
from pathlib import Path

base = Path(".")
orders = pd.read_csv(base / "orders.csv")
users = pd.read_csv(base / "users.csv")

merged = pd.merge(orders, users, on="user_id")
print(merged[["username", "amount"]])
```

```python
import pandas as pd

today_orders = pd.DataFrame({
    "order_id": [1, 2],
    "item": ["Pizza", "Burger"],
    "price": [250, 120]
})

yesterday_orders = pd.DataFrame({
    "order_id": [3, 4],
    "item": ["Pasta", "Sandwich"],
    "price": [200, 90]
})

combined = pd.concat([today_orders, yesterday_orders], ignore_index=True)
print(combined)
```
